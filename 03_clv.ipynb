```python
import pandas as pd
df = pd.read_csv("../data/telco_clean.csv")
df.shape
```

```python
seg = df.groupby("Contract").agg(
    customers=("customerID", "count"),
    avg_monthly=("MonthlyCharges", "mean"),
    churn_rate=("ChurnFlag", "mean")
)
seg["lifetime_months"] = (12 / seg["churn_rate"]).clip(upper=60)
seg["CLV"] = seg["avg_monthly"] * seg["lifetime_months"]
seg.round(2)
```

```python
risk = df[df["ChurnFlag"] == 1].groupby("Contract")["MonthlyCharges"].sum() * 12
risk.round(0)
```
